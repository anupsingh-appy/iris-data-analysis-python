```python
import pandas as pd

# Load dataset
df = pd.read_csv('/content/1) iris.csv')

# Show first 5 rows
df.head()
```

```python
# Check dataset information
df.info()
```

```python
# Check statistical summary
df.describe()
```

```python
# Check duplicate rows
df.duplicated().sum()
```

```python
# Count flower species
df['species'].value_counts()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create count plot
sns.countplot(x='species', data=df)

# Add title
plt.title('Distribution of Iris Species')

# Show plot
plt.show()
```

```python
# Histogram for all numerical columns
df.hist(figsize=(10,8))

# Add spacing
plt.tight_layout()

# Show plots
plt.show()
```

```python
# Boxplot for petal length by species
sns.boxplot(x='species', y='petal_length', data=df)

# Add title
plt.title('Petal Length Distribution by Species')

# Show plot
plt.show()
```

```python
# Scatter plot
sns.scatterplot(
    x='sepal_length',
    y='petal_length',
    hue='species',
    data=df
)

# Add title
plt.title('Sepal Length vs Petal Length')

# Show plot
plt.show()
```

Insights:

1. Setosa species is clearly separated from the other species due to smaller petal lengths.

2. Virginica flowers generally have larger sepal and petal lengths compared to other species.

3. A positive relationship exists between sepal length and petal length.

4. Versicolor and Virginica show slight overlap, while Setosa is distinctly clustered.

```python
# Pairplot visualization
sns.pairplot(df, hue='species')

# Show plot
plt.show()
```

```python
# Correlation matrix
correlation = df.drop('species', axis=1).corr()

# Heatmap
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

# Add title
plt.title('Correlation Heatmap')

# Show plot
plt.show()
```

Insights:
   

1.   As petal length increases, petal width also increases.
2.   Flowers with longer sepals tend to have longer petals.
3.   When sepal width increases, petal length tends to decrease slightly









```python
# Save heatmap image
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

# Save image
plt.savefig('correlation_heatmap.png')

plt.show()
```

Key Insights from Iris Dataset Analysis

1. The dataset contains 150 flower observations equally distributed among three species:
   - Setosa
   - Versicolor
   - Virginica

2. No missing or duplicate values were found, indicating high-quality data.

3. Petal length and petal width show a very strong positive correlation (0.96).

4. Setosa species is clearly separable from the other flower species due to smaller petal dimensions.

5. Virginica flowers generally have the largest petal and sepal measurements.

6. The dataset demonstrates strong relationships between floral dimensions, making it suitable for classification and predictive analysis.
